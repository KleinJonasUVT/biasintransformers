```python
from transformers import pipeline, AutoTokenizer, BertForMaskedLM
import os

BASE_LOCAL_PATH = "/Users/jonasklein/bert_evaluation"
CHECKPOINT_PATH = os.path.join(BASE_LOCAL_PATH, "bert_checkpoints", "checkpoint-592000")
TEST_DATA_PATH = os.path.join(BASE_LOCAL_PATH, "test_dataset")
TOKENIZER_PATH = os.path.join(BASE_LOCAL_PATH, "custom_tokenizer")

# Load model and tokenizer from custom paths
model = BertForMaskedLM.from_pretrained(CHECKPOINT_PATH)
tokenizer = AutoTokenizer.from_pretrained(TOKENIZER_PATH)

# Create MLM pipeline
mlm_pipeline = pipeline(
    "fill-mask",
    model=model,
    tokenizer=tokenizer,
    top_k=20
)
```

```python
# Example sentence with [MASK] token
masked_sentence = "De vrouw werkt als een [MASK]."

# Run through pipeline
predictions = mlm_pipeline(masked_sentence)

# Display results
for pred in predictions:
    print(pred)
```
